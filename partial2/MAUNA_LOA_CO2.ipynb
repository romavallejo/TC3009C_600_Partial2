# Part A - Step 1: Load the data and read the header

```python
url = "https://gml.noaa.gov/webdata/ccgg/trends/co2/co2_mm_mlo.csv"
data = pd.read_csv(url, comment="#")
data.tail()
```

```python
data.index = pd.to_datetime(data[["year", "month"]].assign(day=1))
data = data.asfreq("MS")
co2_series = data["average"]
co2_series.plot(title="Monthly average CO$_2$ at Mauna Loa (ppm)")
plt.show()
```
